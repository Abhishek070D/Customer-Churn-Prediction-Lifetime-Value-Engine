```python
import pandas as pd
import numpy as np
 
df = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

print("Missing TotalCharges:", df["TotalCharges"].isnull().sum())
print("TotalCharges dtype:", df["TotalCharges"].dtype)
```

```python
df[df["TotalCharges"].isnull()][
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
]
```

```python
df = df.dropna(subset=["TotalCharges"])

print("New Dataset Shape:", df.shape)
print("Missing TotalCharges:", df["TotalCharges"].isnull().sum())
```

```python
print("Dataset Shape:", df.shape)
print(df.columns.tolist())
```

```python
categorical_cols = df.select_dtypes(include="object").columns

print("Categorical Columns:")
print(categorical_cols.tolist())

print("\nNumber of Categorical Columns:", len(categorical_cols))
```

```python
print(categorical_cols.tolist())
```

```python
X = df.drop("Churn", axis=1)
y = df["Churn"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
X_model = X.drop("customerID", axis=1)

print("Model Features Shape:", X_model.shape)
print("Model Features:")
print(X_model.columns.tolist())
```

```python
y = y.map({
    "No": 0,
    "Yes": 1
})

print(y.value_counts())
print("\nMissing values:", y.isnull().sum())
```

```python
X_encoded = pd.get_dummies(
    X_model,
    drop_first=True
)

print("Encoded Dataset Shape:", X_encoded.shape)
print("\nFirst 5 rows:")
display(X_encoded.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("X_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape:", X_test_scaled.shape)
```

```python
print("Training data shape:", X_train_scaled.shape)
print("Testing data shape:", X_test_scaled.shape)

print("\nMissing values in training data:",
      np.isnan(X_train_scaled).sum())

print("Missing values in testing data:",
      np.isnan(X_test_scaled).sum())

print("\nTarget distribution:")
print(y.value_counts())
```

# Preprocessing Summary

- Removed 11 rows with missing `TotalCharges` values caused by customers with 0 months of tenure.
- Converted `TotalCharges` from object to numeric.
- Preserved `customerID` for customer-level predictions and reporting.
- Separated `Churn` as the target variable.
- Converted `Churn` into binary values: 0 = No, 1 = Yes.
- Applied One-Hot Encoding to categorical features.
- Created 30 model-ready features.
- Split the dataset into 80% training and 20% testing data using stratification.
- Applied StandardScaler to the training and testing features.
- Verified that there are no missing values in the scaled datasets.
